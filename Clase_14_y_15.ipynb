```python
#*----------------------------------------------------------------*
#  Ejemplo: Generar una calculadora de la Ley de Ohm interactiva  |
#*----------------------------------------------------------------*

import ipywidgets as wid

# Paso 1: definir una función

def calculo_v(I,R):
  v = I*R
  return v

def R_equ_serie(R1, R2):
  R_equ = R1 + R2
  return R_equ

def R_equ_paralelo(R1, R2):
  R_equ = 1/(1/R1 + 1/R2)
  return R_equ


#Paso 2: Generar un entorno que permita modificar los valores de la corriente
# y la resistencia en tiempo real.

#print(calculo_v(2,4)) #----> Valores fijos

resultado = wid.interact(calculo_v,I = wid.FloatSlider(min = 0, max= 50, step = 0.1, description = "Corriente (A)"),
             R = wid.Dropdown(options = [100, 110, 200, 220, 470, 560, 820, 1000, 1200], description = "Resistencia (Ohm)" )
             )

resultado = wid.interact(R_equ_serie,R1 = wid.Dropdown(options = [100, 110, 200, 220, 470, 560, 820, 1000, 1200], description = "Resistencia 1 (Ohm)" ),
             R2 = wid.Dropdown(options = [100, 110, 200, 220, 470, 560, 820, 1000, 1200], description = "Resistencia 2 (Ohm)" )
             )
```

```python
#*------------------------------------------------*
#|  Ejemplo: Imprimir resultados en formato .txt  |
#*------------------------------------------------*

# los documentos txt contienen datos estructurados: Texto, faciles de leer e interpretar para el computador.

# Calcular voltajes y corriente de un circuito mixto: R1 + (R2 || R3)

#Definir el voltaje de entrada

V_fuente = 12 # En voltios

#Calculamos la resistencia equivalente de R2 Y R3 utilizando la función R_equ_paralelo

#                       R1 , R2
R_R_E = R_equ_paralelo(220, 110)

#Calculamos la resistencia equivalente de R1 Y R_euivalente anterior utilizando la función R_equ_serie

#                    R1 ,  R2
R_E_T = R_equ_serie(100, R_R_E)

# Calcular la corriente del circuito

I_total = V_fuente / R_E_T

# Calcular el voltaje en las resistencias:
# Caluclar el voltaje de la resistencia R1:

V_R1 = I_total * 220

# Calcular el voltaje de la resistencia R2 = V_R3

V_R2 = V_fuente - V_R1
V_R3 = V_R2

# Calcular las corrientes

I_R2 = V_R2 / 110
I_R3 = V_R3 / 100


#Definir el contenido del documento (S e recomienda utilizar una lista)

contenido = [

            "*-------------------------------------------------*",
            "|             Resultados del cálculo              |",
            "*-------------------------------------------------*",
            "\n"
            "*-------------------------------------------------*",
            "|                 Datos de entrada                |",
            "*-------------------------------------------------*",

            f"Voltaje de la fuente: {V_fuente} V",
            f"Resistencia 1: {220} Ohms",
            f"Resistencia 2: {110} Ohms",
            f"Resistencia 3: {100} Ohms",
            "\n",
            "*-------------------------------------------------*",
            "|                 Resultados                      |",
            "*-------------------------------------------------*",
            "\n"
            f"Corriente total: {I_total *1000:.2f} mA",
            f"Voltaje de la resistencia 1: {V_R1:.2f} V",
            f"Voltaje de la resistencia 2: {V_R2:.2f} V",
            f"Voltaje de la resistencia 3: {V_R3:.2f} V",
            f"Corriente de la resistencia 2: {I_R2 *1000:.2f} mA",
            f"Corriente de la resistencia 3: {I_R3 * 1000:.2f} mA",
            "\n",

]

# Guardar los resultados en un archivo .txt

nombre_archivo = "Resultados_circuito.txt"

# Le indicamos a la computadora como debe traducir los caracteres.

with open(nombre_archivo, "w", encoding= "utf-8") as archivo:
  for linea in contenido:
    archivo.write(linea + "\n")


```

```python
#*------------------------------------------------*
#|  Ejemplo: Imprimir resultados en formato .xlsx |
#*------------------------------------------------*

import pandas as pd


# los documentos txt contienen datos estructurados: Texto, faciles de leer e interpretar para el computador.

# Calcular voltajes y corriente de un circuito mixto: R1 + (R2 || R3)

#Definir el voltaje de entrada

V_fuente = 12 # En voltios

#Calculamos la resistencia equivalente de R2 Y R3 utilizando la función R_equ_paralelo

#                       R1 , R2
R_R_E = R_equ_paralelo(220, 110)

#Calculamos la resistencia equivalente de R1 Y R_euivalente anterior utilizando la función R_equ_serie

#                    R1 ,  R2
R_E_T = R_equ_serie(100, R_R_E)

# Calcular la corriente del circuito

I_total = V_fuente / R_E_T

# Calcular el voltaje en las resistencias:
# Caluclar el voltaje de la resistencia R1:

V_R1 = I_total * 220

# Calcular el voltaje de la resistencia R2 = V_R3

V_R2 = V_fuente - V_R1
V_R3 = V_R2

# Calcular las corrientes

I_R2 = V_R2 / 110
I_R3 = V_R3 / 100

# Organizar datos dentro del archivo excel

Datos_excel = {
              "Variable": ["Voltaje de la fuente", "Resistencia equivalente total", "V_R1", "V_R2", "V_R3", "I_R2", "I_R3", "I_total"],
              "Valor": [V_fuente, R_E_T, V_R1, V_R2, V_R3, I_total * 1000, I_R2 *1000, I_R3*1000],
              "Unidad": ["V", "Ohm", "V", "V", "V", "mA", "mA", "mA"]
}


# Se organizan los datos calculados en un dataframe

data_frame_datos = pd.DataFrame(Datos_excel)


# Guardar los resultados en un archivo .xlsx

nombre_archivo = "Resultados_circuito.xlsx"

# Le indicamos a la computadora como debe traducir los caracteres.


data_frame_datos.to_excel(nombre_archivo, index = False)

```
